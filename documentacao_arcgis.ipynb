### <span style="color:#336699">Passo 1: Acesso à Página de Sandbox </span>
<hr style="border:1px solid #0077b9;">

O primeiro passo para experimentar a plataforma por um período determinado é acessar o portal oficial de avaliação da Esri. 

> ⚠️ Clique no link abaixo para iniciar o seu cadastro:

Acessar Trial do ArcGIS Online (https://www.esri.com/pt-br/arcgis/products/arcgis-online/trial)

### <span style="color:#336699">Passo 2: Pré-cadastro e Envio do E-mail </span>
<hr style="border:1px solid #0077b9;">

Ao acessar o link, você será redirecionado para a página de inscrição da Esri.

Preencha o formulário com o seu e-mail corporativo, nome e informações da organização.

Após concluir as configurações iniciais, o sistema da Esri enviará automaticamente um e-mail de validação com o seu link exclusivo de acesso.

### <span style="color:#336699">Passo 3: Ativação da Conta</span>
<hr style="border:1px solid #0077b9;">

Acesse a sua caixa de entrada e procure pela mensagem de confirmação. O fluxo final de ativação seguirá o modelo abaixo:

De: ArcGIS Notification notifications@arcgis.com
Assunto: Ativação de Conta Pública do ArcGIS

Obrigado por se registrar em uma Conta Pública do ArcGIS. Você está apenas a uma etapa de iniciar. Clique no link abaixo para ativar sua conta:

 Ativar minha Conta Pública do ArcGIS

> ⚠️ Nota de CS: Este link de ativação expirará em 24 horas.

### <span style="color:#336699">Passo 4:  Acesso ao Painel do ArcGis</span>
<hr style="border:1px solid #0077b9;">

Após acessar o link por e-mail você será redirecionado para o ambiente online da ferramenta.

### <span style="color:#336699">Passo 5:  Conhecendo o Software</span>
<hr style="border:1px solid #0077b9;">

Nesta etapa, você poderá explorar livremente as funcionalidades do software. Para este teste, vamos criar uma camada de dados vetoriais contendo pontos de referência para representar o uso e ocupação do solo.

Siga as etapas abaixo para configurar e desenhar seus dados:

- Configurar a Camada: No menu lateral esquerdo, acesse o painel Camadas. No campo Nome da Feição (ou Título da Camada), insira o nome desejado para identificar o seu tema de estudo.

- Inserir os Pontos: Com a ferramenta de feição de ponto (símbolo) selecionada, navegue pelo mapa e clique diretamente sobre a tela nos locais que deseja mapear para fixar os seus pontos de referência.

- Aproveite este momento para testar a criação de múltiplos pontos e explorar como o software armazena essas localizações!
